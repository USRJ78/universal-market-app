```python
import yfinance as yf
import pandas as pd
import pickle

def load_nse_tickers():
    df = pd.read_csv("data/EQUITY_L.csv")
    return [s + ".NS" for s in df["SYMBOL"].dropna().unique() if "&" not in s]

tickers = load_nse_tickers()

all_data = {}

for ticker in tickers:
    try:
        df = yf.download(ticker, period="10y", progress=False)
        if df.empty:
            continue
        all_data[ticker] = df
    except:
        continue

with open("data_cache.pkl", "wb") as f:
    pickle.dump(all_data, f)

print("✅ Data cached")
```
